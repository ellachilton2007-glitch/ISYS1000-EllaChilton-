## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).


```
create a complete Python program called Smart Purchase Advisor. The program should help a person decide whether they can afford a planned purchase based on their monthly income, current savings, essential monthly expenses and the purchase price.

Ask the user to input:
- monthly after-tax income
- current savings
- Essential monthly expenses
- Purchase price

Do not allow negative numbers, if user enters negative numbers ask them to enter again.

Calculate
- Disposable income = monthly income - essential expenses
- Savings remaining = current savings - purchase price
Purchase price as a percentage of savings
- Purchase price as a percentage of monthly income

Use these rules:
1. DO NOT BUY if the purchase costs more than the person's savings OR their essential expenses are greater than their income
2. SAVE MORE FIRST if the purchase uses more then 50% of savings or costs more then 30% of monthly income
3. Affordable with caution if the purchase uses between 25% and 50% of savings
4. Affordable if none of the previous rules apply.

If monthly income is zero, redommend DO NOT BUy and do not divide zero
After making teh decision, display all the input amounts, the calculated disposable income and remaining savings. teh relevant percentages, the final recommendation, and a short explanation of why that recommensation was given

Round dollar amounts to two decimal places and percentages to one decimal place. make the output clear and easy to read

Use only standard python. The program must run completely without errors.
```

````python
# Paste the code your prompt produced here, then run it to show it works.
````

````python
def get_positive_input(prompt_message):
    while True:
        try:
            value = float(input(prompt_message))
            if value < 0:
                print("Error: Please enter a non-negative number.")
            else:
                return value
        except ValueError:
            print("Error: Invalid input. Please enter a number.")

# Display title and explanation
print("\n--- Smart Purchase Advisor ---")
print("This tool helps you decide if you can afford a planned purchase based on your finances.")
print("-----------------------------\n")

# Get user inputs with validation
monthly_income = get_positive_input("Enter your monthly after-tax income ($): ")
current_savings = get_positive_input("Enter your current savings ($): ")
essential_expenses = get_positive_input("Enter your essential monthly expenses ($): ")
purchase_price = get_positive_input("Enter the price of the planned purchase ($): ")

# Calculate financial metrics
monthly_disposable_income = monthly_income - essential_expenses
savings_remaining_after_purchase = current_savings - purchase_price

# Handle division by zero for income and savings percentages
if monthly_income > 0:
    purchase_percentage_of_income = (purchase_price / monthly_income) * 100
else:
    purchase_percentage_of_income = float('inf') # Effectively makes Rule 1 apply if income is 0

if current_savings > 0:
    purchase_percentage_of_savings = (purchase_price / current_savings) * 100
else:
    purchase_percentage_of_savings = float('inf') # Effectively makes Rule 1 apply if savings is 0 and price > 0

# Determine recommendation based on rules
recommendation = ""
explanation = ""

# Rule 1 — Do Not Buy
if monthly_income == 0 or purchase_price > current_savings or essential_expenses > monthly_income:
    recommendation = "DO NOT BUY"
    if monthly_income == 0:
        explanation = "You have no monthly income available to support this purchase."
    elif purchase_price > current_savings:
        explanation = f"The purchase price (${purchase_price:.2f}) is greater than your current savings (${current_savings:.2f})."
    else:
        explanation = f"Your essential monthly expenses (${essential_expenses:.2f}) exceed your monthly income (${monthly_income:.2f})."

# Rule 2 — Consider Saving More First
elif purchase_percentage_of_savings > 50 or purchase_percentage_of_income > 30:
    recommendation = "SAVE MORE FIRST"
    if purchase_percentage_of_savings > 50:
        explanation = f"The purchase would use {purchase_percentage_of_savings:.1f}% of your current savings, which is more than 50%."
    else:
        explanation = f"The purchase price is {purchase_percentage_of_income:.1f}% of your monthly income, which is more than 30%."

# Rule 3 — Affordable With Caution
elif 25 <= purchase_percentage_of_savings <= 50:
    recommendation = "AFFORDABLE WITH CAUTION"
    explanation = f"The purchase is affordable based on available savings, but it would use {purchase_percentage_of_savings:.1f}% of your current savings."

# Rule 4 — Affordable
else:
    recommendation = "AFFORDABLE"
    explanation = "The purchase is within the tool's affordability limits. It would use less than 25% of current savings and no more than 30% of monthly income."

# Display summary
print("\n--- Financial Summary ---")
print(f"Monthly Income:                  ${monthly_income:,.2f}")
print(f"Current Savings:                 ${current_savings:,.2f}")
print(f"Essential Monthly Expenses:      ${essential_expenses:,.2f}")
print(f"Price of Planned Purchase:       ${purchase_price:,.2f}")
print("---------------------------")
print(f"Monthly Disposable Income:       ${monthly_disposable_income:,.2f}")
print(f"Savings Remaining After Purchase: ${savings_remaining_after_purchase:,.2f}")

if monthly_income > 0:
    print(f"Purchase as % of Monthly Income: {purchase_percentage_of_income:.1f}%")
else:
    print("Purchase as % of Monthly Income: Not applicable (income is zero)")

if current_savings > 0:
    print(f"Purchase as % of Current Savings: {purchase_percentage_of_savings:.1f}%")
else:
    print("Purchase as % of Current Savings: Not applicable (savings is zero)")

print("\n--- Recommendation ---")
print(f"Final Recommendation: {recommendation}")
print(f"Explanation: {explanation}")
print("----------------------")
````
