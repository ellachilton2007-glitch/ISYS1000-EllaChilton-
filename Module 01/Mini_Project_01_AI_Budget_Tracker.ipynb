## Part 3: Building the Budget Tracker

Now for the real thing. Ask AI to help you write a program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative

> 💡 **Pro tip:** The more specific your prompt, the less clean-up you'll need to do afterwards. "Write a budget tracker" gets you something generic; describing the exact inputs, outputs, and edge cases gets you something closer to what you actually need.


```python
# Write your detailed prompt here:
"""
My Budget Tracker Prompt:
Write a Python program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative
"""
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE

try:
    # Get monthly income
    income_str = input("Enter your monthly income: ")
    monthly_income = float(income_str)

    if monthly_income < 0:
        print("Income cannot be negative. Please enter a positive value.")
    else:
        expenses = []
        for i in range(3):
            while True:
                expense_str = input(f"Enter expense {i+1} (or press Enter to skip): ")
                if not expense_str:
                    break
                try:
                    expense = float(expense_str)
                    if expense < 0:
                        print("Expense cannot be negative. Please enter a positive value.")
                    else:
                        expenses.append(expense)
                        break
                except ValueError:
                    print("Invalid input. Please enter a number for the expense.")

        total_expenses = sum(expenses)
        remaining_budget = monthly_income - total_expenses

        print(f"\n--- Budget Summary ---")
        print(f"Monthly Income: ${monthly_income:.2f}")
        print(f"Total Expenses: ${total_expenses:.2f}")
        print(f"Remaining Budget: ${remaining_budget:.2f}")

        if remaining_budget < 0:
            print("🚨 Warning: Your budget is negative! You are spending more than you earn.")
        elif remaining_budget == 0:
            print("📝 Note: Your budget is balanced.")
        else:
            print("✅ Good job! You have a positive remaining budget.")

except ValueError:
    print("Invalid input for income. Please enter a number.")
```

```python
# Document your enhancement choice and the prompt you used
enhancement_choice = "I chose to add: Percentage of income spent"
enhancement_prompt = """
Add a feature to the existing Python budget tracker program that:
- Calculates the percentage of monthly income spent on expenses.
- Prints this percentage in the budget summary.
- Handles cases where monthly income might be zero to avoid division by zero errors.
"""

print(f"🚀 Enhancement: {enhancement_choice}")
```

```python
# ENHANCED BUDGET TRACKER

try:
    # Get monthly income
    income_str = input("Enter your monthly income: ")
    monthly_income = float(income_str)

    if monthly_income < 0:
        print("Income cannot be negative. Please enter a positive value.")
    else:
        expenses = []
        for i in range(3):
            while True:
                expense_str = input(f"Enter expense {i+1} (or press Enter to skip): ")
                if not expense_str:
                    break
                try:
                    expense = float(expense_str)
                    if expense < 0:
                        print("Expense cannot be negative. Please enter a positive value.")
                    else:
                        expenses.append(expense)
                        break
                except ValueError:
                    print("Invalid input. Please enter a number for the expense.")

        total_expenses = sum(expenses)
        remaining_budget = monthly_income - total_expenses

        print(f"\n--- Budget Summary ---")
        print(f"Monthly Income: ${monthly_income:.2f}")
        print(f"Total Expenses: ${total_expenses:.2f}")
        print(f"Remaining Budget: ${remaining_budget:.2f}")

        # Calculate and print percentage of income spent
        if monthly_income > 0:
            percentage_spent = (total_expenses / monthly_income) * 100
            print(f"Percentage of Income Spent: {percentage_spent:.2f}%")
        else:
            print("Percentage of Income Spent: N/A (income is zero or negative)")

        if remaining_budget < 0:
            print("🚨 Warning: Your budget is negative! You are spending more than you earn.")
        elif remaining_budget == 0:
            print("📝 Note: Your budget is balanced.")
        else:
            print("✅ Good job! You have a positive remaining budget.")

except ValueError:
    print("Invalid input for income. Please enter a number.")
```
